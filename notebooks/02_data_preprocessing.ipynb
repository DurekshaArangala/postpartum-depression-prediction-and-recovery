```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("../dataset/raw/post natal data.csv")

# Check dataset
print(df.head())
print(df.shape)
```

Remove irrelevant timestamp

```python
df_clean = df.drop(columns=["Timestamp"])

print(df_clean.shape)
```

Standardize text values

```python
for column in df_clean.columns:

    df_clean[column] = df_clean[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text cleaning completed!")
```

Handle missing values

```python
for column in df_clean.columns:

    if df_clean[column].isnull().sum() > 0:

        mode_value = df_clean[column].mode()[0]

        df_clean[column] = df_clean[column].fillna(mode_value)

print("Missing values handled!")
print(df_clean.isnull().sum())
```
